## Bash command to setup port forwarding

It seems the MYSQL libraries don't make an SSH connection themselves. Instead, you must run seperate code to start an SSH connection and then forward the port to 127.0.0.1:3307. The MYSQL library then queries from this

```bash
ssh -i <path_to_ssh_key> -fN -L 3307:localhost:3306 charlotte@tist.org
```

## Make a connection

````python
#DON'T WRITE THE PASSWORD INTO THE CODE
pwd = input("password")
print('before connection')

#Make sure port fowarding is in place beofore this will work
cnx = connector.connect(user='charlotte', password=pwd,
                        host='127.0.0.1',
                        port = 3307)
print('after connection')

# input('input to exit')
# cnx.close() #Really we should make sure the connection is closed
````
